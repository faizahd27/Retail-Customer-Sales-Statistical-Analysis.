Alpha = 0.05. Results are observational; do not interpret correlation as causation.

```
import pandas as pd, numpy as np
from scipy import stats
import statsmodels.api as sm
df=pd.read_csv("2026M01410048_Cleaned_Dataset.csv")
```

## Descriptive statistics

```
cols=["Sales","Profit","Age","Income"]
summary=df[cols].describe().T
summary["variance"]=df[cols].var()
summary["mode"]=df[cols].mode().iloc[0]
summary["IQR"]=summary["75%"]-summary["25%"]
print(summary.to_string())
```

## 95% confidence interval for mean Sales

```
s=df.Sales.dropna();print("Mean sales CI:",stats.t.interval(.95,len(s)-1,loc=s.mean(),scale=stats.sem(s)))
```

## Welch t-test: North versus South

```
a=df.loc[df.Region=="North","Sales"].dropna();b=df.loc[df.Region=="South","Sales"].dropna();result=stats.ttest_ind(a,b,equal_var=False);print("H0: equal regional means; H1: unequal means",result,"Reject H0?",result.pvalue<.05)
```

## ANOVA: sales across regions

```
result=stats.f_oneway(*[g.Sales.dropna() for _,g in df.groupby("Region")]);print("H0: all regional means equal; H1: at least one differs",result,"Reject H0?",result.pvalue<.05)
```

## Chi-square: customer segment and category

```
ct=pd.crosstab(df.Customer_Segment,df.Product_Category);result=stats.chi2_contingency(ct);print("H0: independent; H1: associated", "chi2",result.statistic,"p",result.pvalue,"dof",result.dof,"Reject H0?",result.pvalue<.05)
```

## Pearson correlation: discount and profit

```
s=df[["Discount","Profit"]].dropna();result=stats.pearsonr(s.Discount,s.Profit);print("H0: correlation zero; H1: nonzero",result,"Reject H0?",result.pvalue<.05,"Correlation is not causation")
```

## Simple linear regression: Sales ~ Quantity

```
s=df[["Sales","Quantity"]].dropna();model=sm.OLS(s.Sales,sm.add_constant(s[["Quantity"]])).fit();print(model.summary())
```

## Multiple linear regression: Sales ~ Quantity + Discount + Income

```
cols=["Quantity","Discount","Income"];s=df[["Sales"]+cols].dropna();model=sm.OLS(s.Sales,sm.add_constant(s[cols])).fit();print(model.summary())
```
